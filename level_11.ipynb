```python
from pyspark.sql import SparkSession

# Create Spark session
spark = SparkSession.builder \
    .appName("PySpark Example") \
    .getOrCreate()

# Create sample data
data = [
    ("Alice", 25),
    ("Bob", 30),
    ("Cathy", 29)
]

# Define columns
columns = ["Name", "Age"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Show DataFrame
df.show()

# Filter data
df_filtered = df.filter(df.Age > 27)
df_filtered.show()
```
